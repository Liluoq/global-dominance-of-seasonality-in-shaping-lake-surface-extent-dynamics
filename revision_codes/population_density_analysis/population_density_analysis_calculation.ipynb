```python
import pandas as pd
import geopandas as gpd
import numpy as np
from datetime import datetime
from dateutil.relativedelta import relativedelta
import sys
sys.path.append('../../my_spatial_analyze/data_analyze/basin_wise_analysis')
import basin_wise_analysis

lake_lse_pkl_path = '/WORK/Data/global_lake_area/area_csvs/lakes/pkl/lakes_all_with_all_additional_attributes_revision_with_all_added_analysis.pkl'
lake_lse_gdf = pd.read_pickle(lake_lse_pkl_path)
lake_lse_gdf_save_path = '/WORK/Data/global_lake_area/area_csvs/lakes/pkl/lakes_all_with_all_additional_attributes_revision_with_all_added_analysis_finals.pkl'

basinatlas_shp_path = '/WORK/Data/global_lake_area/hydroATLAS/shp/BasinATLAS_v10_shp/BasinATLAS_v10_lev06.shp'
basinatlas_gdf = gpd.read_file(basinatlas_shp_path)

hydrolakes_pnt_path = '/WORK/Data/global_lake_area/hydroATLAS/hydroLAKES/HydroLAKES_points_v10_shp/HydroLAKES_points_v10.gpkg'
hydrolakes_pnt_gdf = gpd.read_file(hydrolakes_pnt_path)

assert basinatlas_gdf.crs == hydrolakes_pnt_gdf.crs, 'BasinATLAS and HydroLAKES CRSs are not the same.'

hydrolakes_with_population_density = basinatlas_gdf.sjoin(hydrolakes_pnt_gdf, how='inner', predicate='contains')[['Hylak_id', 'ppd_pk_sav']]

lake_lse_gdf = lake_lse_gdf.merge(hydrolakes_with_population_density, on='Hylak_id', how='left')

lake_lse_gdf.to_pickle(lake_lse_gdf_save_path)
```
