```python
import pandas as pd
import geopandas as gpd
import sys
from datetime import datetime
from dateutil.relativedelta import relativedelta
sys.path.append('../../my_spatial_analyze/data_analyze/extreme_analysis')
import area_extreme_analysis

lake_lse_gdf_path = '/WORK/Data/global_lake_area/area_csvs/lakes/pkl/lakes_all_with_all_additional_attributes_revision_with_high_water_extreme_flag_and_changes_and_percentage.pkl'
lake_lse_gdf_save_path = '/WORK/Data/global_lake_area/area_csvs/lakes/pkl/lakes_all_with_all_additional_attributes_revision_with_all_added_analysis.pkl'
lake_lse_gdf = pd.read_pickle(lake_lse_gdf_path)

output_column_prefix = 'increases_in_low_water_extremes'

area_start_date = '2001-01-01'
area_end_date = '2024-01-01'
date_fmt = '%Y-%m-%d'
area_start_date = datetime.strptime(area_start_date, date_fmt)
area_end_date = datetime.strptime(area_end_date, date_fmt)
current_date = area_start_date
area_columns = []
while current_date < area_end_date:
    area_columns.append(current_date.strftime(date_fmt))
    current_date += relativedelta(months=1)
area_mask_columns = [f'frozen_{col}' for col in area_columns]
low_water_extreme_flag_columns = [f'low_water_extreme_{col}' for col in area_columns]

area_first_period_start_date = '2001-01-01'
area_first_period_end_date = '2012-01-01'
area_second_period_start_date = '2013-01-01'
area_second_period_end_date = '2024-01-01'
area_first_period_start_date = datetime.strptime(area_first_period_start_date, date_fmt)
area_first_period_end_date = datetime.strptime(area_first_period_end_date, date_fmt)
area_second_period_start_date = datetime.strptime(area_second_period_start_date, date_fmt)
area_second_period_end_date = datetime.strptime(area_second_period_end_date, date_fmt)
area_columns_first_period = []
current_date = area_first_period_start_date
while current_date < area_first_period_end_date:
    area_columns_first_period.append(current_date.strftime(date_fmt))
    current_date += relativedelta(months=1)
area_columns_second_period = []
current_date = area_second_period_start_date
while current_date < area_second_period_end_date:
    area_columns_second_period.append(current_date.strftime(date_fmt))
    current_date += relativedelta(months=1)
area_mask_columns_first_period = [f'frozen_{col}' for col in area_columns_first_period]
area_mask_columns_second_period = [f'frozen_{col}' for col in area_columns_second_period]
low_water_extreme_flag_columns_first_period = [f'low_water_extreme_{col}' for col in area_columns_first_period]
low_water_extreme_flag_columns_second_period = [f'low_water_extreme_{col}' for col in area_columns_second_period]

lake_lse_gdf[[f'{output_column_prefix}_frequency', f'{output_column_prefix}_count', f'{output_column_prefix}_pval']] = area_extreme_analysis.increases_in_low_water_extremes_and_pval_of_two_periods(
    lake_lse_df_with_low_water_extreme_flags=lake_lse_gdf,
    first_period_low_water_extreme_flag_columns=low_water_extreme_flag_columns_first_period,
    first_period_area_mask_columns=area_mask_columns_first_period,
    second_period_low_water_extreme_flag_columns=low_water_extreme_flag_columns_second_period,
    second_period_area_mask_columns=area_mask_columns_second_period,
    output_column_prefix=output_column_prefix,
)
lake_lse_gdf.to_pickle(lake_lse_gdf_save_path)
```

```python
import pandas as pd
import geopandas as gpd
import numpy as np
from datetime import datetime
from dateutil.relativedelta import relativedelta
import sys
sys.path.append('/WORK/Codes/global_lake_area/my_spatial_analyze/')
import grid_analyze

lake_lse_pkl_path = '/WORK/Data/global_lake_area/area_csvs/lakes/pkl/lakes_all_with_all_additional_attributes_revision_with_all_added_analysis.pkl'
lake_lse_gdf = pd.read_pickle(lake_lse_pkl_path)

area_start_date = '2001-01-01'
area_end_date = '2024-01-01'
date_fmt = '%Y-%m-%d'
area_start_date = datetime.strptime(area_start_date, date_fmt)
area_end_date = datetime.strptime(area_end_date, date_fmt)
area_columns = []
current_date = area_start_date
while current_date < area_end_date:
    area_columns.append(current_date.strftime(date_fmt))
    current_date = current_date + relativedelta(months=1)


additional_agg_dict_for_generating_grid = {
    'increases_in_high_water_extremes_count': ('mean', 'median'),
    'increases_in_high_water_extremes_frequency': ('mean', 'median'),
    'increases_in_low_water_extremes_count': ('mean', 'median'),
    'increases_in_low_water_extremes_frequency': ('mean', 'median'),
}

lake_lse_gdf = lake_lse_gdf[['Hylak_id'] + list(additional_agg_dict_for_generating_grid.keys()) + area_columns + ['centroid', 'geometry', 'Lake_type']]


grid_gdf = grid_analyze.generate_grid_from_geometry_added_concatenated_areas(
    geometry_added_concatenated_areas_gdf=lake_lse_gdf,
    grid_size=0.5,
    area_columns=area_columns,
    grid_extent=[-180, -90, 180, 90],
    geometry_to_use_column='geometry',
    additional_agg_dict=additional_agg_dict_for_generating_grid,
    bootstrap_ks_test_column_pairs=None,
    bootstrap_ks_test_output_columns=None,
    verbose=1
)

grid_gdf_save_path = '/WORK/Data/global_lake_area/area_csvs/grids/pkl/grid_all_medium_for_extreme_changes.pkl'
grid_gdf = grid_gdf[grid_gdf['lake_count'] != 0]
grid_gdf.to_pickle(grid_gdf_save_path)
```
