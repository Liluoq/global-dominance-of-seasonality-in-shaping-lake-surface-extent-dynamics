```python
import pandas as pd
import geopandas as gpd
from datetime import datetime
from dateutil.relativedelta import relativedelta

lake_lse_gdf_path = '/WORK/Data/global_lake_area/area_csvs/lakes/pkl/lakes_all_with_all_additional_attributes_revision_with_all_added_analysis_finals.pkl'
lake_lse_gdf_1 = pd.read_pickle(lake_lse_gdf_path)

area_start_date = '2001-01-01'
area_end_date = '2024-01-01'
date_fmt = '%Y-%m-%d'
area_start_date = datetime.strptime(area_start_date, date_fmt)
area_end_date = datetime.strptime(area_end_date, date_fmt)
current_date = area_start_date
area_columns = []
while current_date < area_end_date:
    area_columns.append(current_date.strftime(date_fmt))
    current_date += relativedelta(months=1)
area_mask_columns = [f'frozen_{col}' for col in area_columns]
high_water_extreme_flag_columns = [f'high_water_extreme_{col}' for col in area_columns]
high_water_extreme_property_columns = [
    'increases_in_high_water_extremes_frequency', 
    'increases_in_high_water_extremes_count', 
    'increases_in_high_water_extremes_pval',
    'increases_in_low_water_extremes_frequency', 
    'increases_in_low_water_extremes_count', 
    'increases_in_low_water_extremes_pval',
    'max_deviations_of_high_water_extremes_from_annual_means_second_period',
    'percentage_of_extreme_high_water_compared_with_long_term_trends_plus_average_seasonlity'
    ]

population_density_columns = ['ppd_pk_sav']

attributes_columns = ['Hylak_id'] + high_water_extreme_flag_columns + high_water_extreme_property_columns + population_density_columns

lake_lse_gdf_1 = lake_lse_gdf_1[attributes_columns].copy()

lake_lse_gdf_2_path = '/WORK/Data/global_lake_area/area_csvs/lakes/pkl/lakes_all_with_all_additional_attributes_revision_with_missing_data_and_unbuffered_boundary_area_and_total_variation_relative_to_total_area.pkl'
lake_lse_gdf_2 = pd.read_pickle(lake_lse_gdf_2_path)

# merge lake_lse_gdf_1 and lake_lse_gdf_2 on Hylak_id
lake_lse_gdf = pd.merge(lake_lse_gdf_1, lake_lse_gdf_2, on='Hylak_id', how='inner')



```

```python
lake_lse_gdf.to_csv('/WORK/Data/global_lake_area/area_csvs/lakes/csv/monthly_lake_surface_extent_with_all_analysis.csv', index=False)
```
